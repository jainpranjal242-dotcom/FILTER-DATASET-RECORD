## Objective
Filter records using one or more Boolean conditions.

```python
import pandas as pd

df=pd.read_csv('titanic_sample.csv')
df.head()
```

## Example 1 — Age greater than 30
Select passengers older than 30.

```python
filtered_age=df[df['age']>30]
print(filtered_age)
print('Number of records:',len(filtered_age))
```

## Example 2 — Survived passengers
Select records where survived equals 1.

```python
filtered_survived=df[df['survived']==1]
print(filtered_survived)
print('Number of records:',len(filtered_survived))
```

## Example 3 — Multiple conditions with AND
Select passengers older than 30 who survived. Parentheses are used around each condition.

```python
filtered_and=df[(df['age']>30) & (df['survived']==1)]
print(filtered_and)
print('Number of records:',len(filtered_and))
```

## Example 4 — Multiple conditions with OR
Select passengers with fare above 30 OR age below 18.

```python
filtered_or=df[(df['fare']>30) | (df['age']<18)]
print(filtered_or)
print('Number of records:',len(filtered_or))
```

## Example 5 — Categorical filtering
Select passengers who boarded at Cherbourg.

```python
filtered_city=df[df['embarked']=='Cherbourg']
print(filtered_city)
print('Number of records:',len(filtered_city))
```

## Key Observations
1. Boolean conditions return matching rows.
2. `&` means AND.
3. `|` means OR.
4. Parentheses are important around conditions.
5. `len()` or `.shape[0]` checks the resulting number of records.
6. Numeric and categorical columns can both be filtered.

## Conclusion
Pandas filtering is useful for retrieving relevant records during data exploration and preprocessing.
